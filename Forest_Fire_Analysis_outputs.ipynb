```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
```

```python
uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_excel(file_name, sheet_name="Forest_Fire_Data")

print("Dataset shape (rows, columns):", df.shape)
print(df.columns.tolist())
display(df.head())
```

```python
import pandas as pd
import numpy as np

print(df.head())
```

```python

df.info()

display(df.select_dtypes(include="number").describe())

print("Missing values by column:")
print(df.isna().sum())
```

```python
#Q1 — Calculate average fire incidence
fire_col = "Fire_Frequency_Count"

average_fire_incidence = df[fire_col].mean()

print("Average Fire Incidence per Region-Month:",
      round(average_fire_incidence, 2))

print("Note: This is a simulated educational dataset.")
```

```python
#Q2 — Filter data for a selected region
selected_region = "California, USA"

region_data = df[df["Region"] == selected_region]

display(region_data)

print("Total records for", selected_region, ":", len(region_data))
```

```python
#Q3 — Data Cleaning is completed successfully
# Create a copy of the original dataset
clean_df = df.copy()

# Columns containing missing values
numeric_columns = [
    "Fire_Frequency_Count",
    "Estimated_Burned_Area_km2"
]

# Fill missing values using the median of each region
for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(
        clean_df.groupby("Region")[column].transform("median")
    )

    # If a region has no median, use the overall median
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )

print("Missing values after cleaning:")
print(clean_df.isna().sum())
```

```python
#Q4 — Grouping and Aggregation
# Group fire data by season
season_summary = clean_df.groupby(
    "Hemisphere_Season"
).agg(
    Average_Fire_Frequency=("Fire_Frequency_Count", "mean"),
    Average_Burned_Area=("Estimated_Burned_Area_km2", "mean")
).reset_index()

display(season_summary.round(2))
```

```python

region_summary = clean_df.groupby(
    ["Continent", "Region"]
).agg(
    Total_Fire_Frequency=("Fire_Frequency_Count", "sum"),
    Total_Burned_Area=("Estimated_Burned_Area_km2", "sum")
).reset_index()

display(region_summary.head(10).round(2))
```

```python
#Q5 — Data Visualization using graphs.

import matplotlib.pyplot as plt
import seaborn as sns

monthly_summary = clean_df.groupby(
    ["Month_Number", "Month"], as_index=False
).agg(
    Average_Fire_Frequency=("Fire_Frequency_Count", "mean")
).sort_values("Month_Number")

plt.figure(figsize=(10, 5))
sns.lineplot(
    data=monthly_summary,
    x="Month",
    y="Average_Fire_Frequency",
    marker="o"
)
plt.title("Monthly Forest Fire Frequency Trend")
plt.xlabel("Month")
plt.ylabel("Average Fire Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(9, 5))

sns.barplot(
    data=clean_df,
    x="Hemisphere_Season",
    y="Fire_Frequency_Count",
    errorbar=None
)

plt.title("Average Fire Frequency by Season")
plt.xlabel("Season")
plt.ylabel("Average Fire Frequency")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

top_regions = (
    clean_df.groupby("Region")["Fire_Frequency_Count"]
    .sum()
    .nlargest(10)
    .reset_index()
)

plt.figure(figsize=(11, 6))
sns.barplot(
    data=top_regions,
    x="Fire_Frequency_Count",
    y="Region"
)

plt.title("Top 10 Regions by Total Fire Frequency")
plt.xlabel("Total Fire Frequency")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

```python

clean_df.to_csv("cleaned_forest_fire_data.csv", index=False)

print("Cleaned dataset saved successfully!")

```

```python

season_summary.to_csv("seasonal_fire_summary.csv", index=False)

print("Seasonal summary saved successfully!")
```
